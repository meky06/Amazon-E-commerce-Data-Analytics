```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

Loading DataFrame

```python
customer_data = pd.read_csv('Training.csv')
testing_data = pd.read_csv('testing.csv')
```

```python
x = customer_data.drop(columns=['Repeat_Purchase','customer_unique_id'] , axis=1)
y = customer_data['Repeat_Purchase']
```

```python
test = testing_data.drop(columns='customer_unique_id' , axis=1)
```

```python
x.isna().sum()
```

```python
x["Avg_Late_Delivery_Days"] = x["Avg_Late_Delivery_Days"].fillna(0)
```

```python
testing_data["Avg_Late_Delivery_Days"] = testing_data["Avg_Late_Delivery_Days"].fillna(0)
```

```python
model = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)
model.fit(x, y)
```

```python
x_training_prediction = model.predict(x)
training_data_accuracy = accuracy_score(x_training_prediction, y)
print('Accuracy on training data : ', training_data_accuracy)
```

```python
y_pred = model.predict(test)
y_prob = model.predict_proba(test)[:, 1]
```

```python
testing_data["Predicted_Repeated_Purchase"] = y_pred
testing_data["Repeated_Purchase_Probability"] = y_prob
```

```python
testing_data.head()
```

```python
testing_data.to_csv(
    "Amazon_Repeated_Purchase_Predictions.csv",
    index=False
)
```
